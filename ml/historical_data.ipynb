```python
import fastf1
import pandas as pd
import numpy as np
import time
```

```python
seasons = [2021, 2022, 2023, 2024, 2025]
```

```python
schedule = fastf1.get_event_schedule(2025)

schedule[
    ["RoundNumber", "EventName", "EventDate"]
].head()
```

```python
all_results = []

for year in seasons:
    print(f"\nCollecting {year} season...")
    
    schedule = fastf1.get_event_schedule(year)
    
    for _, event in schedule.iterrows():
        
        round_number = event["RoundNumber"]
        event_name = event["EventName"]
        
        # Skip testing/non-race events
        if pd.isna(round_number) or round_number == 0:
            continue
        
        print(f"  Loading Round {int(round_number)}: {event_name}")
        
        try:
            session = fastf1.get_session(
                year,
                int(round_number),
                "R"
            )
            
            session.load(
                laps=False,
                telemetry=False,
                weather=False,
                messages=False
            )
            
            results = session.results.copy()
            
            results["season"] = year
            results["round"] = int(round_number)
            results["event"] = event_name
            
            all_results.append(results)
            
            time.sleep(1)
            
        except Exception as e:
            print(f"    Failed: {e}")
```

```python
historical_results = pd.concat(
    all_results,
    ignore_index=True
)
```

```python
historical_results.shape
historical_results.head()
```

```python
historical_results.columns.tolist()
```

```python
dataset = historical_results[
    [
        "season",
        "round",
        "event",
        "Abbreviation",
        "FullName",
        "TeamName",
        "GridPosition",
        "Position",
        "Points"
    ]
].copy()
```

```python
dataset.rename(
    columns={
        "Abbreviation": "driver",
        "FullName": "driver_name",
        "TeamName": "team",
        "GridPosition": "grid_position",
        "Position": "race_position",
        "Points": "race_points"
    },
    inplace=True
)
```

```python
dataset["winner"] = (
    dataset["race_position"] == 1
).astype(int)
```

```python
dataset["podium"] = (
    dataset["race_position"] <= 3
).astype(int)
```

```python
qualifying_results = []

for year in seasons:
    print(f"\nCollecting qualifying data for {year}...")
    
    schedule = fastf1.get_event_schedule(year)
    
    for _, event in schedule.iterrows():
        
        round_number = event["RoundNumber"]
        
        if pd.isna(round_number) or round_number == 0:
            continue
        
        try:
            session = fastf1.get_session(
                year,
                int(round_number),
                "Q"
            )
            
            session.load(
                laps=False,
                telemetry=False,
                weather=False,
                messages=False
            )
            
            results = session.results.copy()
            
            results["season"] = year
            results["round"] = int(round_number)
            
            qualifying_results.append(results)
            
            time.sleep(1)
            
        except Exception as e:
            print(f"Failed: {year} round {round_number}: {e}")
```

```python
qualifying = pd.concat(
    qualifying_results,
    ignore_index=True
)
```

```python
qualifying = qualifying[
    [
        "season",
        "round",
        "Abbreviation",
        "Position"
    ]
].copy()
```

```python
qualifying.rename(
    columns={
        "Abbreviation": "driver",
        "Position": "qualifying_position"
    },
    inplace=True
)
```

```python
dataset = dataset.merge(
    qualifying,
    on=["season", "round", "driver"],
    how="left"
)
```

```python
dataset.to_csv(
    "../data/f1_historical_results_2021_2025.csv",
    index=False
)
```
